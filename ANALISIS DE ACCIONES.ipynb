```python
import time
import webbrowser
import pyperclip
import pyautogui
import urllib.parse

# Definimos las variables
destinatario = "ale.franco.valencia@gmail.com"
asunto = "Analisis de acciones ultimos 6 meses"

mensaje = (
    "Buenas tardes, le envio el analisis de acciones de los ultimos 6 meses para su revision.\n\n"
    f"Cotización Máxima:\n{maxima.to_string()}\n\n"
    f"Cotización Mínima:\n{minima.to_string()}\n\n"
    f"Cotización Media:\n{valormedio.to_string()}\n\n"
    "Quedo atento a sus comentarios, saludos cordiales."
)

# Codificamos el asunto para que sea seguro en una URL
asunto_codificado = urllib.parse.quote(asunto)

# 1. Abrimos Gmail directamente con destinatario y asunto ya escritos usando un enlace directo de Gmail
enlace_gmail = f"https://mail.google.com/mail/?view=cm&fs=1&to={destinatario}&su={asunto_codificado}"
webbrowser.open(enlace_gmail)

# 2. Esperamos a que cargue la página web por completo (ajusta los segundos si tu internet es lento)
print("Abriendo el navegador... Esperando 8 segundos a que cargue la interfaz...")
time.sleep(8)

# 3. Al abrir ese enlace, Gmail enfoca automáticamente el cuerpo del correo o el asunto.
# Copiamos nuestro gran mensaje con las métricas al portapapeles
pyperclip.copy(mensaje)

# 4. Hacemos un solo clic en una zona central del navegador para asegurar que la ventana está activa
# Nota: Ajusta las coordenadas si es necesario, o simplemente haz clic tú mismo en la zona del texto apenas cargue.
print("Haz clic manualmente en el cuerpo del mensaje si el cursor no aparece automáticamente.")

# Intentamos hacer clic automático en la zona de redacción y pegar
time.sleep(2)
pyautogui.hotkey('ctrl', 'v')

print("¡Proceso terminado!")
```
